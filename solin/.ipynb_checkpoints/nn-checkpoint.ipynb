```python
import oemof.solph as solph
import pandas as pd
from oemof.solph import create_time_index
```

```python
df = pd.read_csv(
        "data.csv",
        index_col=0,
        parse_dates=True
    )
```

```python
timeindex = create_time_index(2025)
es = solph.EnergySystem(timeindex=timeindex)
el_bus = solph.Bus(label="ele")
heat_bus = solph.Bus(label="heat")
```

```python
es.add(el_bus)
```

```python
grid = solph.components.Source(
        label="el_grid",
        outputs={
            el_bus: solph.flows.Flow(
                variable_costs= df["el_price_eur_kwh"],
                
            )
        },
    )
#pv = solph.components.Source(
 #       label="pv_dach",
 #       outputs={
  #          el_bus: solph.flows.Flow(
   #             nominal_capacity=df["pv_power_kw"].max(),
   #             fix= df["pv_power_kw"]/df["pv_power_kw"].max(),
                
                
  #          )
   #     },
  #  )
```

```python
#hp = solph.components.Converter(
    label="heat_pump",
    inputs={el_bus: solph.flows.Flow()},
    outputs={
        heat_bus: solph.flows.Flow(
            nominal_capacity=900,   # kW Wärmeleistung max
        )
    },
   
)
```

```python
max_demand = df["heat_demand_kw"].max()
demand = solph.components.Sink(
        label="last",
        inputs={
            el_bus: solph.flows.Flow(
                nominal_capacity=df["heat_demand_kw"].max(),
                fix=df["heat_demand_kw"]/max_demand,
            )
        },
    )
```

```python
es.add( grid,demand)
```

```python
print([n.label for n in es.nodes])
```

```python
model =  solph.Model(es)
model.solve(solver="cbc", solve_kwargs={"tee": True})
```
